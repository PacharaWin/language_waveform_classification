```python
def validSquare(p1, p2, p3, p4):
        def find_dist(point1, point2):
            return (point1[0]**2-point2[0]**2) + (point1[1]**2 - point2[1]**2)

        points = [p1, p2, p3, p4]
        dists = []
        
        for i in range(4):
            for j in range(i+1, 4):
                dist = find_dist(points[i], points[j])
                dists.append(dist)

        dists.sort()

        return (dists[0] == dists[1] == dists[2] == dists[3] and
        dists[0] > 0 and dists[4] == dists[5] and dists[4] == 2*dists[1])
```

```python
validSquare(p1 = [1,0], p2 = [-1,0], p3 = [0,1], p4 = [0,-1])
```

```python
def inc(input):
    ans = []
    flag = False
    for i in input:
        if i.isdigit():
            if flag:
                ans[-1] = ans[-1] + i
            else:
                ans.append(i)
            flag = True
        else:
            flag = False
    for i in range(len(ans)):
        ans[i] = int(ans[i])
    print(ans)
    if ans == sorted(ans):
        return True
    return False
            

inc("1 box has 3 blue 4 red 6 green and 12 yellow marbles")
```
