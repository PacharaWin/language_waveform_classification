```python
import pandas as pd
import matplotlib as plt
```

```python
df = pd.read_pickle("RomanceDigitData.pkl")
df.head()
```

```python
df['language'].value_counts()
```

```python
df[df['language']=='PO']['gender'].value_counts()
```

```python
# Group by language and gender, then count the unique speaker IDs
speaker_counts = df.groupby(['language', 'gender'])['speaker'].nunique()

# Unstack the data to make it look like a clean table (Pivot)
speaker_distribution_table = speaker_counts.unstack(fill_value=0)

print("=== UNIQUE SPEAKERS BY LANGUAGE & GENDER ===")
print(speaker_distribution_table)

# If you want to see the total unique speakers per language as well:
print("\n=== TOTAL UNIQUE SPEAKERS PER LANGUAGE ===")
print(speaker_distribution_table.sum(axis=1))
```
